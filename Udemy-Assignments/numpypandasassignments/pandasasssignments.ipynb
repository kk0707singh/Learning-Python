```python
# question4.1
catagory = np.random.choice(['a','b','c'], size=10)
value = np.random.randint(1,10, size=10)
df = pd.DataFrame({
    'Catagory':catagory,
    'Value':value
})
print(df)
grouped = df.groupby('Catagory')
result = grouped['Value'].agg(['sum', 'mean'])
print('result\n', result)
```

```python
# question4.2
product = np.random.choice(['Product1','Product2','Product3'], size=10)
catagory = np.random.choice(['a','b','c'], size=10)
sales = np.random.randint(1,10, size=10)
df = pd.DataFrame({
    'product':product,
    'Catagory':catagory,
    'Sales':sales
})
print(df)

grouped = df.groupby('Catagory')
result = grouped['Sales'].sum()
print('result\n',result)
```

```python
# question5.1
import pandas as pd
import numpy as np
df1 = pd.DataFrame({
    'Key': ['A', 'B', 'C', 'D'],
    "Value": np.random.randint(1,11, size=4)
})
print('df1 is:\n', df1)

df2 = pd.DataFrame({
    'Key': ['A', 'B', 'C', 'E'],
    'Value': np.random.randint(1,11, size=4)
})
print("df2 is:\n", df2)

merged_col = pd.merge(df1,df2, on="Key")
print('merged columns are:\n', merged_col)
```

```python
# question5.2
df1 = pd.DataFrame({
    'a': np.random.randint(1,11, size=4),
    'b': np.random.randint(1,11, size=4)    
})

df2 = pd.DataFrame({
    'c': np.random.randint(1,11, size=4),
    'd': np.random.randint(1,11, size=4)    
})
print(f'df1 is:\n{df1}\ndf2 is:\n{df2}')

row_conct = pd.concat([df1, df2], axis=0)
print('row concatination:\n',row_conct)

col_conct = pd.concat([df1, df2], axis=1)
print('column concatination:\n', col_conct)
```

```python
# question6.1
date = pd.date_range(start='2025-01-01', end='2025-07-07', freq='D')
print('date\n',date)

values = np.random.randint(1,51, size=len(date))
print('value:\n',values)

df = pd.DataFrame({'Values': values}, index=date)
print('df is:\n',df)

# Resample the DataFrame to compute the monthly mean of the values
monthly_mean = df.resample('M').mean()
print("Monthly mean DataFrame:")
print(monthly_mean)
```

```python
# question6.2
date = pd.date_range(start='2021-01-01', end='2021-12-31', freq='D')
# print('date\n',date)

values = np.random.randint(1,51, size=len(date))
# print('value:\n',values)

df = pd.DataFrame({'Date': date, 'Values': values})
df.set_index('Date', inplace=True)
print('df is:\n',df)

# oneway to find rolling mean by creating a new col:
df['rolling_mean'] = df['Values'].rolling(window=7).mean()
print(df)
```

```python
# question7.1
# Create a Pandas DataFrame with a MultiIndex (hierarchical index)
arrays = [['A', 'A', 'B', 'B'], ['one', 'two', 'one', 'two']]
index = pd.MultiIndex.from_arrays(arrays, names=('Category', 'SubCategory'))
df = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)), index=index, columns=['Value1', 'Value2', 'Value3'])
print("MultiIndex DataFrame:")
print(df)

# Basic indexing and slicing operations
print("Indexing at Category 'A':")
print(df.loc['A'])

print("Slicing at Category 'B' and SubCategory 'two':")
print(df.loc[('B', 'two')])
```

```python
# question7.2
# Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'
arrays = [['A', 'A', 'B', 'B', 'C', 'C'], ['one', 'two', 'one', 'two', 'one', 'two']]
index = pd.MultiIndex.from_arrays(arrays, names=('Category', 'SubCategory'))
df = pd.DataFrame(np.random.randint(1, 100, size=(6, 3)), index=index, columns=['Value1', 'Value2', 'Value3'])
print("MultiIndex DataFrame:")
print(df)

# Compute the sum of values for each 'Category' and 'SubCategory'
sum_values = df.groupby(['Category', 'SubCategory']).sum()
print("Sum of values:")
print(sum_values)
```
