```python
# question6.1
data = np.random.randint(1,6, size=(3,3))
print(f'Original matrix:\n {data}')

# computing determinant:
determinant = np.linalg.det(data)
print(f'Determinant is: {determinant}')

# computing inverse:
inverse = np.linalg.inv(data)
print(f'Inverse of data:\n {inverse}')

# computing eighen value
eighen_val = np.linalg.eigvals(data)
print(f'Eighen Value is:\n {eighen_val}')
```

```python
# question10.1
import numpy.ma as ma
data = np.random.randint(1,20, size=(4,4))
print('original data\n', data)
masked_data = ma.masked_greater(data, 10)
print(masked_data)
print(f'masked data sum:\n {masked_data.sum()}')
```

```python
# question10.2
import numpy.ma as ma
data = np.random.randint(1,10, size=(3,3))
print('original data\n', data)
masked_data = ma.masked_array(data, np.eye(3, dtype=bool))
print(masked_data)

mean_unmasked = masked_data.mean()
print(mean_unmasked)
masked_data = masked_data.filled(mean_unmasked)
print(masked_data)
```
