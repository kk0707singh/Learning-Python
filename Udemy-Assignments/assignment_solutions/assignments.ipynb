```python
# question8:
count = 0
user_num = []
num = int(input("Enter the no"))
while num!=0:
    user_num.append(num)
    count = count+num
    num = int(input("Enter the no"))
print(count)
print(user_num)
```

```python
# question11:
num = int(input("Enter the no"))
even_num = []
for i in range(1, num+1):
    if(i%2==0):
        even_num.append(i)
print(even_num)
```

```python
# question12:
num = int(input("Enter the no"))
fact = 1
for i in range(1, num+1):
    fact = fact*i
print(f"factorial result is: {fact}")
```

```python
# Question13: 
num = int(input("Enter the no"))
sum = 0

while num!=0:
    digit = num%10
    sum = sum+digit
    num = num//10
print(sum)
```
